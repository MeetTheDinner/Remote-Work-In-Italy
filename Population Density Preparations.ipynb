Values imported as strings. Need to be recoded to integer types.

```python
    
def transform_data(dataframe):
    dataframe.columns = dataframe.columns.str.strip()
    dataframe = dataframe.rename(columns=lambda x: x.strip())

    province_mapping = {
        "Bolzano / Bozen": "Bolzano-Bozen",
        "Bolzano": "Bolzano-Bozen",
        "Forlï¿½-Cesena": "Forlì-Cesena",
        "Monza e della Brianza": "Monza-Brianza",
        "Pesaro e Urbino": "Pesaro Urbino",
        "Reggio di Calabria": "Reggio Calabria",
        "Reggio nell'Emilia": "Reggio Emilia",
        "Valle d'Aosta": "Aosta",
        "Valle d'Aosta / Vallï¿½e d'Aoste": "Aosta",
        "Valle d'Aosta / Vallée d'Aoste": "Aosta",
        "Valle d'Aosta": "Aosta",
        }
    """
        "Olbia-Tempio" : "Sassari",
        "Ogliastra" : "Nuoro",
        "Medio Campidano" : "Sud Sardegna",
        "Carbonia-Iglesias" : "Sud Sardegna" """

    if dataframe.columns[0] in dataframe.columns:
        dataframe[dataframe.columns[0]] = dataframe[dataframe.columns[0]].replace(province_mapping)
    
    return dataframe

df_area = transform_data(df_area)
df_pop = transform_data(df_pop)
```

```python
def transform_float(dataframe):
    for col in dataframe.columns[1:]:
            dataframe[col] = dataframe[col].astype(str).str.strip().str.replace(",", "", regex=False)
            dataframe[col] = dataframe[col].astype(str).str.strip().str.replace(".", "", regex=False)
            dataframe[col] = pd.to_numeric(dataframe[col], errors='coerce').fillna(0).astype(float)
    return dataframe

def transform_int(dataframe):
    for col in dataframe.columns[1:]:
            dataframe[col] = dataframe[col].astype(str).str.strip().str.replace(",", "", regex=False)
            dataframe[col] = dataframe[col].astype(str).str.strip().str.replace(".", "", regex=False)
            dataframe[col] = pd.to_numeric(dataframe[col], errors='coerce').fillna(0).astype(int)
    return dataframe
```

```python
df_area = transform_int(df_area)
df_pop = transform_int(df_pop)
```

```python
df_pop.info()
df_area.info()
```

```python
for col in df_area.columns[1:]:
    df_area[col] = (df_area[col] / 10000).round().astype(int)
```

```python
df_area.info()
df_pop.info()
```

```python
df_area.head()
```

```python
# build a mapping once
area_map = df_area.set_index("Indicator")["Total area (km2)"]

df_density = df_pop.copy()

for col in df_pop.columns[1:]:
    # for every year column, divide the population by the area of the
    # province in the same row
    df_density[col] = df_density[col] / df_pop["Province"].map(area_map)


df_density = df_density.round(2)

df_density.info()
```
